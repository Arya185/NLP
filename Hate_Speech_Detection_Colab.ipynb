```python
import os
import pandas as pd
import numpy as np

# Let's create dummy/sample English & Hinglish datasets so the code can execute end-to-end safely
# In a real run, these will be loaded from Hugging Face 'datasets' library or CSVs.

os.makedirs('data', exist_ok=True)

# 1. Davidson-like dataset (English)
en_data = {
    'text': [
        'I absolutely love this product!',
        'Go back to where you came from, you are not wanted here.',
        'This is a normal neutral statement about the weather.',
        'Stupid idiots should be banned from this platform. I hate them.',
        'Hope you have a beautiful day ahead!'
    ] * 20,
    'label': [0, 1, 0, 1, 0] * 20  # 0: Neutral/Normal, 1: Hate Speech/Toxic
}
pd.DataFrame(en_data).to_csv('data/english_dataset.csv', index=False)

# 2. HASOC-like Hinglish dataset (Hinglish: Hindi written in Latin alphabet mixed with English)
hinglish_data = {
    'text': [
        'Yeh product bohot achha hai!',
        'Tum log pagal ho, desh chhod ke bhaag jao.',
        'Aaj ka mausam kaisa hai bhai?',
        'F**k off, tumhari himmat kaise hui ye bolne ki?',
        'Aapka din shubh ho!'
    ] * 20,
    'label': [0, 1, 0, 1, 0] * 20  # 0: Non-Toxic, 1: Toxic
}
pd.DataFrame(hinglish_data).to_csv('data/hinglish_dataset.csv', index=False)

print("Local demo datasets successfully generated!")
```

```python
from sklearn.model_selection import train_test_split

# Load datasets
df_en = pd.read_csv('data/english_dataset.csv')
df_hi = pd.read_csv('data/hinglish_dataset.csv')

# Split English Dataset
train_en, val_en = train_test_split(df_en, test_size=0.2, random_state=42, stratify=df_en['label'])
# Split Hinglish Dataset
train_hi, val_hi = train_test_split(df_hi, test_size=0.2, random_state=42, stratify=df_hi['label'])

print(f"English Dataset: Train size={len(train_en)}, Val size={len(val_en)}")
print(f"Hinglish Dataset: Train size={len(train_hi)}, Val size={len(val_hi)}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Fit TF-IDF Vectorizer on English Dataset
vectorizer_en = TfidfVectorizer(max_features=5000)
X_train_en = vectorizer_en.fit_transform(train_en['text'])
X_val_en = vectorizer_en.transform(val_en['text'])

# Train Baseline Model
baseline_model_en = LogisticRegression()
baseline_model_en.fit(X_train_en, train_en['label'])

# Evaluate Baseline
preds_en = baseline_model_en.predict(X_val_en)
print("=== English Baseline Evaluation ===")
print(classification_report(val_en['label'], preds_en))
```

```python
# Example layout for Transformers (can run on CPU or T4 GPU if activated in Runtime)
import torch
from transformers import AutoTokenizer

model_name = 'google/muril-base-cased'
try:
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    print(f"Successfully loaded tokenizer for {model_name}")

    # Tokenization sample check
    inputs = tokenizer("Yeh project bohot badiya hai!", return_tensors="pt")
    print("Tokenized output keys:", list(inputs.keys()))
except Exception as e:
    print("Error loading transformer:", e)
```
